## "break" and "continue" statements
- break is used to exit a for loop or a while loop, whereas continue is used to skip the current block, and return to the "for" or "while" statement.

```python
# Prints out 0,1,2,3,4

count = 0
while True:
    print(count)
    count += 1
    if count >= 5:
        break

# Prints out only odd numbers - 1,3,5,7,9
for x in range(10):
    # Check if x is even
    if x % 2 == 0:
        continue
    print(x)
```

## Can we use "else" clause for loops?
Unlike languages like C,CPP.. we can use else for loops. When the loop condition of "for" or "while" statement fails then code part in "else" is executed. If a break statement is executed inside the for loop then the "else" part is skipped. Note that the "else" part is executed even if there is a continue statement.

```python
# Prints out 0,1,2,3,4 and then it prints "count value reached 5"

count=0
while(count<5):
    print(count)
    count +=1
else:
    print("count value reached %d" %(count))

# Prints out 1,2,3,4
for i in range(1, 10):
    if(i%5==0):
        break
    print(i)
else:
    print("this is not printed because for loop is terminated because of break but not due to fail in condition")
```

In Python, self represents the specific instance (the object) of the class. It acts as a reference to the exact object you are currently working with, allowing you to access and modify its attributes and methods.

Why is self Required?
Unlike some other programming languages (like C++, Java, or JavaScript) that handle this implicitly via keywords like this, Python requires you to explicitly pass the instance as the first parameter to any method inside a class.

When you create multiple objects from the same class, Python needs a way to know which object's data you want to read or change.

How It Works Behind the Scenes
When you call a method on an object, Python automatically passes that object into the method's self parameter.

For example, consider this class:

Python


class Dog:
    def __init__(self, name):
        self.name = name  # 'self' binds the name to the specific object

    def bark(self):
        print(f"{self.name} says woof!")

 Creating two different objects
dog1 = Dog("Buddy")
dog2 = Dog("Max")

 When you call this:
dog1.bark()

Python translates it behind the scenes to this:
Dog.bark(dog1)
Without self, Python would have no way of knowing whether bark() should print "Buddy says woof!" or "Max says woof!". It ensures that the method operates on the correct instance data.

(Note: You can technically name the parameter anything you want—like this or me—but sticking to self is a universally accepted Python convention that every developer expects.)

## COnstructor 
That specific method is called the constructor (or initializer) in Python.
def __init__(self, name):
    self.name = name
What It Does
When you create a new object from a class, Python automatically runs the __init__ method. It acts as the setup crew or a blueprint template for every new object (like dog1 or dog2).

__init__: Short for "initialize." It prepares the object right when it is born.

self: Refers to the specific new object being created (e.g., dog1).

name: An argument you pass in when you create the object (e.g., "Buddy").

self.name = name: Takes the name you passed in and attaches it to that specific object as an attribute so it can be remembered and used later.

How It Works in Practice
When you write:


dog1 = Dog("Buddy")
Python does two things behind the scenes:

Creates a blank new Dog object in memory.

Automatically calls Dog.__init__(dog1, "Buddy"), which sets dog1.name = "Buddy".

If you later create a second dog:


dog2 = Dog("Max")
It runs the exact same template, but assigns dog2.name = "Max". This ensures every dog object gets its own unique name without mixing them up.

```python
class NumberHolder:

   def __init__(self, number):
       self.number = number

   def returnNumber(self):
       return self.number

var = NumberHolder(7)
print(var.returnNumber()) #Prints '7'
```

## Dictionaries
A dictionary is a data type similar to arrays, but works with keys and values instead of indexes. Each value stored in a dictionary can be accessed using a key, which is any type of object (a string, a number, a list, etc.) instead of using its index to address it.

```python
phonebook = {}
phonebook["John"] = 938477566
phonebook["Jack"] = 938377264
phonebook["Jill"] = 947662781
print(phonebook)
```

```python
phonebook = {
    "John" : 938477566,
    "Jack" : 938377264,
    "Jill" : 947662781
}
print(phonebook)
```

Dictionaries can be iterated over, just like a list. However, a dictionary, unlike a list, does not keep the order of the values stored in it. To iterate over key value pairs, use the following syntax:Programming

```python
phonebook = {"John" : 938477566,"Jack" : 938377264,"Jill" : 947662781}
for name, number in phonebook.items():
    print("Phone number of %s is %d" % (name, number))
```
